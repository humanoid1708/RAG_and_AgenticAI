```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
```

```python
llm = ChatOllama(model="llama3.1:8b")
```

```python
#Invoke
response = llm.invoke("Explain machine learning in one sentence.")

print(response.content)
```

```python
#Prompt Template
prompt = PromptTemplate.from_template(
    "What is the color of {object}?"
)

formatted_prompt = prompt.invoke({
    "object" : "apple"
})

response_prompt = llm.invoke(formatted_prompt)

print(response_prompt.content)
```

```python
#Chaining Concept
chain = prompt | llm

response_chain = chain.invoke({
    "object" : "sky"
})

print(response_chain.content)
```

The response showed above, is an AIMessage object containing these main fields:
| Parameter            | Meaning                                                   |
| -------------------- | --------------------------------------------------------- |
| `content`            | The actual text generated by the LLM                      |
| `additional_kwargs`  | Extra provider-specific data, usually empty               |
| `response_metadata`  | Detailed information about how the response was generated |
| `id`                 | Unique ID for this LangChain run/message                  |
| `tool_calls`         | Tools the LLM requested to execute                        |
| `invalid_tool_calls` | Tool calls that could not be parsed correctly             |
| `usage_metadata`     | Input, output, and total token counts                     |




```python
#LLM Chaining
meal_chain = PromptTemplate.from_template(
    "Name one classic dish from {location}. Give only the name."
) | llm

recipe_chain = PromptTemplate.from_template(
    "Give a short and simple recipe for {meal}."
) | llm

time_chain = PromptTemplate.from_template(
    "Estimate the cooking time for this recipe: {recipe}"
) | llm

meal = meal_chain.invoke({"location": "USA"}).content
recipe = recipe_chain.invoke({"meal": meal}).content
time = time_chain.invoke({"recipe": recipe}).content

print("Meal:", meal)
print("\nRecipe:", recipe)
print("\nTime:", time)
```
